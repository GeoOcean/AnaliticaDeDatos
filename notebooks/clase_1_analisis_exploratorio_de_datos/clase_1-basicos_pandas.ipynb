## Introducción a Pandas

**Pandas** es una librería de Python diseñada para trabajar con **datos estructurados**, como tablas y series de datos.

Sus principales estructuras son:

* **Series**: una secuencia de datos asociada a un índice.
* **DataFrame**: una tabla formada por filas y columnas.

Pandas permite cargar, explorar, seleccionar, filtrar y analizar datos de forma sencilla. Además, se integra fácilmente con otras librerías como **NumPy** y **Matplotlib**.


```python
import pandas as pd
import numpy as np
```

Una Series de Pandas es una estructura que representa una única columna de datos junto con un índice.

```python
#### Series

temperatura = pd.Series([20, 21, 23, 22, 19])

print(temperatura)
```

```python
print(temperatura.mean())
print(temperatura.std())
```

¿Qué es un DataFrame?

Un DataFrame es la estructura de datos principal de Pandas para trabajar con datos organizados en forma de tabla, con filas y columnas.

Cada columna de un DataFrame es una Series, y todas las columnas comparten el mismo índice.

```python
df = pd.DataFrame({
    "temperatura": [20, 21, 23, 22, 19],
    "humedad": [70, 65, 60, 62, 75]
})

print(df)
```

```python
data = {
    "temperatura": [20, 21, 23, 22, 19],
    "humedad": [70, 65, 60, 62, 75]
}

df = pd.DataFrame(data)

print(df)
```

Algunas herramientas útiles para investigar un dataframe:

```python
df.head() #### muestra las primeras filas del DataFrame
```

```python
df.shape  #### muestra el número de filas y columnas del DataFrame
```

```python
df.columns #### muestra los nombres de las columnas del DataFrame
```

```python
df["temperatura"] ### muestra la columna "temperatura" del DataFrame, recupera el objeto Series correspondiente a esa columna
```

Aqui os dejo como unir dos dataframes diferentes, (tienen que compartir indice, si no necesitareis mas tratamiento), esto es util en el proceso de recopilación de datos

```python
np.random.seed(42)

temperatura = np.random.normal(20, 3, 100)
humedad = np.random.normal(60, 10, 100)

df = pd.DataFrame({
    "temperatura": temperatura,
    "humedad": humedad
})
```

```python
df1 = pd.DataFrame({
    "temperatura": [20, 21, 22],
    "humedad": [60, 65, 70]
})

df2 = pd.DataFrame({
    "temperatura": [23, 24, 25],
    "humedad": [68, 64, 61]
})
```

```python
df = pd.concat([df1, df2])

print(df)
```

```python
df = pd.concat([df1, df2], ignore_index=True)
```
