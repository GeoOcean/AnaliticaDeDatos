```python
## Sección 2: Seleccionar y Filtrar Datos

#### Puedes seleccionar una columna de varias formas:
cement = concrete_data['Cement (component 1)(kg in a m^3 mixture)']
print(type(cement))  # Una columna es una Series
print(cement.head())
```

```python
#### Puedes seleccionar varias columnas a la vez usando una lista
selected_cols = concrete_data[['Cement (component 1)(kg in a m^3 mixture)', 'compresive_strenght']]
selected_cols.head()
```

```python
#### Filtrar filas según una condición (WHERE en SQL)
#### Por ejemplo, solo las filas donde la resistencia del concreto es mayor a 40
strength_col = 'compresive_strenght'
strong_concrete = concrete_data[concrete_data[strength_col] > 40]
print(f"Número de muestras con resistencia > 40: {len(strong_concrete)}")
print(f"De un total de: {len(concrete_data)}")
strong_concrete.head()
```

```python
#### Puedes combinar múltiples condiciones con & (AND) y | (OR)
#### Por ejemplo: resistencia entre 30 y 40
strength_col = 'compresive_strenght'
medium_concrete = concrete_data[(concrete_data[strength_col] > 30) & (concrete_data[strength_col] < 40)]
print(f"Muestras con resistencia entre 30 y 40: {len(medium_concrete)}")
medium_concrete.head()
```

```python
## Sección 3: Operaciones Estadísticas y Cálculos

#### Calcular estadísticas sobre una columna
cement_col = concrete_data['Cement (component 1)(kg in a m^3 mixture)']

print(f"Media: {cement_col.mean():.2f}")
print(f"Mediana: {cement_col.median():.2f}")
print(f"Desviación estándar: {cement_col.std():.2f}")
print(f"Mínimo: {cement_col.min():.2f}")
print(f"Máximo: {cement_col.max():.2f}")
print(f"Suma: {cement_col.sum():.2f}")
```

```python
#### Crear una nueva columna a partir de operaciones
#### Por ejemplo, normalizamos la resistencia (restar media y dividir por desv. est.)
strength_col = 'compresive_strenght'
concrete_data['strength_normalized'] = (concrete_data[strength_col] - concrete_data[strength_col].mean()) / concrete_data[strength_col].std()

concrete_data[[strength_col, 'strength_normalized']].head()
```

```python
#### Ordenar (sort) los datos por una columna
sorted_by_strength = concrete_data.sort_values('compresive_strenght', ascending=False)
print("Top 5 muestras con mayor resistencia:")
sorted_by_strength[['Cement (component 1)(kg in a m^3 mixture)', 'compresive_strenght']].head()
```

```python
## Sección 4: Agrupaciones con GroupBy

#### Crear categorías de resistencia para demostrar groupby
def categorize_strength(value):
    if value < 20:
        return 'Baja'
    elif value < 40:
        return 'Media'
    else:
        return 'Alta'

concrete_data['strength_category'] = concrete_data['compresive_strenght'].apply(categorize_strength)

#### Contar cuantas muestras hay en cada categoría
print("Distribución de categorías:")
concrete_data['strength_category'].value_counts()
```

```python
#### Agrupar por categoría y calcular estadísticas
grouped = concrete_data.groupby('strength_category')['Cement (component 1)(kg in a m^3 mixture)'].agg(['mean', 'std', 'count'])
print("Estadísticas de Cemento por categoría de resistencia:")
grouped
```

```python
## Sección 5: Manejo de Valores Faltantes (Missing Values)

#### Ver si hay valores nulos/faltantes en nuestros datos
print("Valores faltantes por columna:")
concrete_data.isnull().sum()

# También podemos verlo como porcentaje
print("\nPorcentaje de valores faltantes:")
(concrete_data.isnull().sum() / len(concrete_data) * 100).round(2)
```

```python
#### Formas de tratar los valores faltantes:

# 1. Eliminar filas con valores nulos
df_without_nulls = concrete_data.dropna()
print(f"Filas originales: {len(concrete_data)}")
print(f"Filas sin nulos: {len(df_without_nulls)}")

# 2. Rellenar valores nulos con un valor específico (ej: media)
# df_filled = concrete_data.fillna(concrete_data.mean())

# 3. Rellenar con el valor anterior o siguiente (útil para series temporales)
# df_filled = concrete_data.fillna(method='ffill')  # forward fill
# df_filled = concrete_data.fillna(method='bfill')  # backward fill
```

## 📚 Resumen de Funciones Principales en Pandas

### **Exploración de Datos**
- `.head()` - Ver las primeras filas
- `.tail()` - Ver las últimas filas
- `.info()` - Información sobre el dataframe
- `.describe()` - Estadísticas descriptivas
- `.shape` - Dimensiones (filas, columnas)

### **Acceso a Datos**
- `df['columna']` - Acceder a una columna (Series)
- `df[['col1', 'col2']]` - Acceder a varias columnas
- `df.iloc[posicion]` - Acceso por posición
- `df.loc[indice]` - Acceso por etiqueta del índice

### **Filtrado**
- `df[df['columna'] > valor]` - Filtrar por condición
- `df[(df['col1'] > val) & (df['col2'] < val)]` - Múltiples condiciones

### **Operaciones Estadísticas**
- `.mean()`, `.median()`, `.std()`, `.min()`, `.max()`, `.sum()`
- `.value_counts()` - Contar valores únicos
- `.groupby()` - Agrupar datos
- `.agg()` - Agregar múltiples funciones

### **Manipulación**
- `df.sort_values()` - Ordenar datos
- `df.rename()` - Renombrar columnas
- `df.isnull()` - Detectar valores faltantes
- `df.dropna()` - Eliminar valores faltantes
- `df.fillna()` - Rellenar valores faltantes

### **Exportar**
- `.to_csv()` - Guardar como CSV
- `.to_excel()` - Guardar como Excel
- `.to_pickle()` - Guardar como Pickle
