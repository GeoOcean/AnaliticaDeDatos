# Introducción a NumPy

**NumPy** (*Numerical Python*) es una librería de Python diseñada para trabajar de forma eficiente con **datos numéricos** y realizar operaciones matemáticas.

Su estructura fundamental es el **array**, una colección ordenada de elementos que permite almacenar y manipular datos de forma eficiente. Los arrays de NumPy son especialmente útiles cuando tenemos grandes cantidades de datos y queremos realizar operaciones sobre ellos sin tener que trabajar elemento a elemento.

NumPy proporciona, entre otras funcionalidades:

* Creación y manipulación de arrays.
* Operaciones matemáticas y estadísticas.
* Generación de números aleatorios.
* Operaciones sobre vectores y matrices.
* Herramientas para trabajar con datos multidimensionales.

En este notebook veremos algunos de los conceptos básicos de NumPy que utilizaremos posteriormente para el **análisis de datos**.


```python
import numpy as np
```

Un np.array es la estructura de datos principal de NumPy. Permite almacenar una colección de valores de forma ordenada y realizar operaciones matemáticas sobre ellos de manera eficiente.

```python
x = np.array([1, 2, 3, 4, 5])
```

```python
print(x)
```

```python
print(x.shape)
print(x.size)
```

Los array pueden ser multidimensionales, por ejemplo:

```python
x_2 = np.array([[1, 2, 3], [4, 5, 6]])
```

```python
print(x_2)
print(x_2.shape)
print(x_2.size)
```

Nos permiten realizar de forma sencilla operaciones vectoriales

```python
#### Operaciones vectoriales
print(x + 10)
print(x * 2)
print(x ** 2)
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])
```

```python
print(x + y) #### No puedo operar si tienen distintas dimensiones
print(x * y)
```

Tambien nos permite obtener estadisticos descriptivos de un array

```python
print(np.mean(x))
print(np.std(x))
print(np.min(x))
print(np.max(x))
print(np.percentile(x, 50))
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 8, 10])

np.corrcoef(x, y)
```

Podemos acceder a los valores de un array por su índice

```python
print(x_2)
```

```python
print(x_2[0])
print(x_2[1, 2])
print(x_2[-1, -1])
```

```python
print(x)
```

```python
print(x[1:3])
```

Python nos ofrece varias formas de interar sobre los valores de un array

```python
for value in x:
    print(value)
```

```python
for i in range(len(x)):
    print("When i is", i, "the value is", x[i])
```

Esto es util para ciertas operaciones, pero siempre sera más eficiente evitar los bucles

```python
x = np.array([1, 2, 3, 4, 5])
for i in range(len(x)):
    x[i] = x[i] * 2

print(x)
```

```python
x = np.array([1, 2, 3, 4, 5])
# Con NumPy
x = x * 2

print(x)
```

Una herramienta muy util es generar datos aleatorios que podeis usar para probar vuestros codigos y modelos. Aqui abajo os dejo como.

```python
np.random.seed(42) #### Cuando trabajeis con variables aleatorias, es recomendable fijar la semilla para que los resultados sean reproducibles. Probad a cambiar la semilla
x = np.random.normal(10, 2, 100)

print(x)
```
